```python
import json
import chromadb
from openai import OpenAI

# Configuración
client = OpenAI()
chroma_client = chromadb.PersistentClient(path="./mi_base_de_datos")
collection = chroma_client.get_or_create_collection(name="faq_cv")

def cargar_faq_desde_json(archivo_json):
    with open(archivo_json, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    for i, item in enumerate(data):
        # Generar embedding
        res = client.embeddings.create(
            input=item['pregunta'], 
            model="text-embedding-3-small"
        )
        embedding = res.data[0].embedding
        
        # Guardar en Chroma
        collection.add(
            ids=[str(i)],
            embeddings=[embedding],
            metadatas=[{"respuesta": item['respuesta']}],
            documents=[item['pregunta']]
        )
    print(f"Se cargaron {len(data)} preguntas exitosamente.")

# Ejecutar
cargar_faq_desde_json('faq.json')
```
